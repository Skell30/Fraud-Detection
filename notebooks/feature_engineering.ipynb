```python
ip['lower_bound_ip_address'] = ip['lower_bound_ip_address'].astype(float).astype(int)
ip['upper_bound_ip_address'] = ip['upper_bound_ip_address'].astype(float).astype(int)

```

```python
ip['lower_bound_ip_address'] = ip['lower_bound_ip_address'].apply(ipaddress.IPv4Address).astype(int)
ip['upper_bound_ip_address'] = ip['upper_bound_ip_address'].apply(ipaddress.IPv4Address).astype(int)
```

```python
fraud.head()
```

```python
def map_country(ip_val):
    row = ip[(ip['lower_bound_ip_address'] <= ip_val) & (ip['upper_bound_ip_address'] >= ip_val)]
    return row['country'].values[0] if not row.empty else 'Unknown'

fraud['country'] = fraud['ip_int'].apply(map_country)
```

```python
fraud['signup_time'] = pd.to_datetime(fraud['signup_time'])
fraud['purchase_time']=pd.to_datetime(fraud['purchase_time'])
```

```python
fraud['time_since_signup'] = (fraud['purchase_time'] - fraud['signup_time']).dt.total_seconds() / 3600
```
